```python
features = [
    "Country Code",
    "City",
    "Locality",
    "Longitude",
    "Latitude",
    "Cuisines",
    "Average Cost for two",
    "Currency",
    "Has Table booking",
    "Has Online delivery",
    "Is delivering now",
    "Switch to order menu",
    "Price range",
    "Votes"
]

target = "Aggregate rating"

X = df[features]
y = df[target]

print("Features selected:", len(features))
print("Target:", target)
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("Selected features:")
for feature in features:
    print("-", feature)
```

```python
numerical_features = [
    "Country Code",
    "Longitude",
    "Latitude",
    "Average Cost for two",
    "Price range",
    "Votes"
]

categorical_features = [
    "City",
    "Locality",
    "Cuisines",
    "Currency",
    "Has Table booking",
    "Has Online delivery",
    "Is delivering now",
    "Switch to order menu"
]

print("Numerical features:", len(numerical_features))
print("Categorical features:", len(categorical_features))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

print("Preprocessing libraries imported successfully.")
```

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

print("Numerical preprocessing pipeline created successfully.")
```

```python
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

print("Categorical preprocessing pipeline created successfully.")
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Complete preprocessing pipeline created successfully.")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
from sklearn.ensemble import RandomForestRegressor

print("Random Forest Regressor imported successfully.")
```

```python
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

print("Model pipeline created successfully.")
```

```python
model.fit(X_train, y_train)

print("Model training completed successfully.")
```

```python
y_pred = model.predict(X_test)

print("Predictions generated successfully.")
print("Number of predictions:", len(y_pred))
```

```python
comparison = pd.DataFrame({
    "Actual Rating": y_test.values,
    "Predicted Rating": y_pred
})

print(comparison.head(10))
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)

print("Mean Squared Error (MSE):", mse)
```

```python
rmse = np.sqrt(mse)

print("Root Mean Squared Error (RMSE):", rmse)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R-squared (R²):", r2)
```

```python
print("RANDOM FOREST REGRESSION RESULTS")
print("-----------------------------------")
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred, alpha=0.5)

plt.xlabel("Actual Rating")
plt.ylabel("Predicted Rating")
plt.title("Actual vs Predicted Restaurant Ratings")

plt.show()
```

```python
feature_names = model.named_steps["preprocessor"].get_feature_names_out()

print("Number of processed features:", len(feature_names))
```

```python
importances = model.named_steps["regressor"].feature_importances_

print("Number of feature importances:", len(importances))
```

```python
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance.head(15))
```

```python
top_features = feature_importance.head(15)

display(top_features)
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"][::-1],
    top_features["Importance"][::-1]
)

plt.xlabel("Feature Importance")
plt.ylabel("Feature")
plt.title("Top 15 Features Influencing Restaurant Ratings")

plt.tight_layout()
plt.show()
```
